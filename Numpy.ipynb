```
import pandas as pd
import matplotlib. pyplot as plt
%matplotlib inline
data = {'Days': ['monday','tuesday', 'wednesday', 'thursday','friday', 'saturday'],
        'week 1': [12,34,67,97,55,44],
        'week 2': [34,56,78,54,34,56],
        'week 3': [56,77,88,65,44,33]}
df = pd.DataFrame(data)
df
```

```
week1 = plt.plot(df['Days'], df['week 1'])
week2 = plt.plot(df['Days'], df['week 2'])
week3 = plt.plot(df['Days'], df['week 3'])
plt.xlabel('Days')
plt.ylabel('temperature')
plt.title('Line plot of week 1 temperature')
plt.legend(['week 1', 'week 2', 'week 3'])
plt.show()

```

```
week1 = plt.plot(df['Days'], df['week 1'], c='y', ls='-.')
week2 = plt.plot(df['Days'], df['week 2'], c='g', ls='dashed')
week3 = plt.plot(df['Days'], df['week 3'], c='b')
plt.xlabel('Days')
plt.ylabel('temperature')
plt.title('Line plot of week 1 temperature')
plt.legend(['week 1', 'week 2', 'week 3'])
plt.show()
```

```
week1 = plt.plot(df['Days'], df['week 1'], c='y', ls='-.')
week2 = plt.plot(df['Days'], df['week 2'], c='g', ls='dashed')
week3 = plt.plot(df['Days'], df['week 3'], c='b', lw=3)   # lw to highligt the line
plt.xlabel('Days')
plt.ylabel('temperature')
plt.title('Line plot of week 1 temperature')
plt.legend(['week 1', 'week 2', 'week 3'])
plt.show()
```
