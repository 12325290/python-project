```python
import seaborn as sns
sns.get_dataset_names()
```

```python
df = sns.load_dataset('titanic')
df
df.head()
```

```python
sns.scatterplot(data = df, x='age', y='fare')
```

```python
sns.scatterplot(data = df, x='age', y='fare', palette = ["red" , "green" , "black"] , hue = "class" )
```

```python
sns.scatterplot(data = df, x='age', y='fare', size = "pclass")
```

```python
########### HISTOGRAM OR DISTRIBUTION PLOT
```

```python
sns.histplot(data = df, x='age')
```

```python
# sns.histplot(data = df, x='age' , bins =15)
sns.histplot(df['age'], bins = 6)                         ## both are same
```

```python
sns.histplot(df['age'], kde = True)
```

```python
##### BARPLOT
```

```python
sns.barplot(x = df['sex'], y = df['fare'], palette=["red", "green"], hue = df['class'])
```

```python
#########  BOX PLOT
```

```python
sns.boxplot(x = df['pclass'], y = df['fare'], palette=["red", "green"])
```

```python
sns.boxplot(x = df['pclass'], y = df['fare'], palette=["red", "green"], hue = df['sex'])
```

```python
##### VIOLIN PLOT
```

```python
sns.violinplot(x = df['pclass'], y = df['fare'], palette=["red", "green"])
```

```python
##### STRIP PLOT
```

```python
sns.stripplot(x = df['pclass'], y = df['fare'], palette=["red", "green"], hue = df['sex'], dodge=True)
```

```python
#### JOINT PLOT
```

```python
sns.jointplot(x = df['age'], y = df['fare'], palette=["red", "green"], hue = df["sex"], kind = "kde")
```

```python
sns.jointplot(x = df['age'], y = df['fare'], palette=["red", "green"], hue = df["sex"], kind = "kde", fill = True)
```

```python
sns.jointplot(x = df['age'], y = df['fare'], palette=["red", "green"], hue = df["sex"], kind = "scatter")
```

```python
from typing import ChainMap
sns.jointplot(x = df['age'], y = df['fare'], palette=["red", "green"], kind = "kde", fill = True, cmap = "Reds")
```

```python
sns.jointplot(x = df['age'], y = df['fare'], palette=["red", "green"], kind = "hex", cmap = "YlGnBu")
```
